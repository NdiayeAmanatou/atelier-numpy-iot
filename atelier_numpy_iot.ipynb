### Partie 1 : Listes Python vs tableaux NumPy 

#### 1) Créer une liste Python stockant ces températures; 

```python
import numpy as np
temperatures = [24.5, 25.1, 26.3, 27.0, 25.8 ]
print (temperatures)
```

#### 2) Créer un tableau NumPy équivalent stockant ces températures ;

```python
temperatures = np.array([24.5, 25.1, 26.3, 27.0, 25.8 ])
print(temperatures)
```

#### 3) Effectuer une multiplication par 2 sur ces deux collections; 

---pour la liste---

```python
mult_liste_temp = temperatures * 2
print(mult_liste_temp)
```

---pour le tableau---

```python
mult_tableau_temp = temperatures * 2
print (mult_tableau_temp)
```

#### 4) Quelle différence observe-t-on ? 

Pour la liste : nous constatons qu'il y'a une duplication de la liste ,les valeurs sont repetes deux fois .
Et pour le tableau : nous constatons un nouveau tableau avec des valeurs multipliees chacune par 2.

### Partie 5 – Filtrage booléen 

##### 1) Afficher les températures supérieures à 30 °C 

```python
temp_sup_30 = temp_journee[temp_journee > 30]
print(temp_sup_30)
```

##### 2) Afficher les températures « normales » càd comprises entre 20 °C et 30 °C 

```python
temp_normale = temp_journee[(temp_journee>20) & (temp_journee<30)]
print(temp_normale)
```

##### 3) Supposons qu'une température supérieure à 35 °C soit considérée comme anormale. 
Déterminer anomalies, un tableau de booléen précisant les anomalies 

```python
anomalies = temp_journee > 35
print (anomalies)
```

***explication***: on a un resultat d'un tableau booleen dont les "True" representent les anomalies c'est a dire les temperatures superieures a 35 et les "False" representent les temperatures normales c'est a dire les temperatures inferieures ou egales a 35.

##### 4) Déterminer le nombre de températures anormales à l’aide de la variable anomalies 

```python
nb_anormales = np.sum(anomalies)
print(nb_anormales)
```

***explication*** : En Python et NumPy, les valeurs booléennes sont traitées numériquement : ce qui fait que True vaut 1 et False vaut 0.
Ainsi, la fonction sum(anomalies) additionne tous les 1 (les True), ce qui donne exactement le nombre total de températures supérieures à 35 °C.

### Partie 6 – Manipulation des dimensions 

##### 1) Créer aléatoirement la matrice représentant les températures mesurées par ces capteurs durant une journée 

```python
rng = np.random.default_rng(seed=42)
matrice_temp = rng.integers(20,45,size=(5,24))
print(matrice_temp)
```

##### 2) Transformer la matrice en un tableau unidimensionnel

```python
matrice_unidim = matrice_temp.flatten()
print(matrice_unidim)
```

##### 3) Déterminer la transposée de la matrice 

```python
matrice_transposee = np.transpose(matrice_temp)
print(matrice_transposee)
```

***explication*** : Dans la matrice initiale matrice_temp, la ligne i contenait les 24 températures relevées par le capteur i.
Dans la matrice transposée matrice_transposee, la ligne j contient maintenant les températures relevées par les 5 capteurs à l'heure j.

##### 4) Extraire toutes les mesures du capteur 1 

```python
mesures_capteur_1 = matrice_temp[0, :]
print(mesures_capteur_1)
```

***explication***: matrice_temp[0, :]
0 (avant la virgule) : permet de sélectionner uniquement la première ligne (le capteur 1) a l'indice 0.
: (après la virgule) : permet de sélectionner l'intégralité des colonnes c'est a dire pour toute les colonnes (les 24 heures de la journée).

##### 5) Extraire toutes les mesures du capteur 5 

```python
mesures_capteur_5 = matrice_temp[4, :]
print(mesures_capteur_5)
```

***explication***: matrice_temp[4, :]
4 (avant la virgule) : permet de sélectionner uniquement la première ligne (le capteur 5) a l'indice 4.
: (après la virgule) : permet de sélectionner l'intégralité des colonnes c'est a dire pour toute les colonnes (les 24 heures de la journée).

##### 6) Extraire les mesures de tous les capteurs à 10h 

```python
mesures_cap = matrice_temp[: ,10]
print(mesures_cap)
```

***explication*** : 
: (avant la virgule) signifie qu'on conserve toutes les lignes (c'est-à-dire l'ensemble des 5 capteurs).
10 (après la virgule) cible uniquement la 11ᵉ colonne (l'index 10 correspondant à la mesure effectuée à 10h).

##### 7) Extraire les mesures des capteurs 2 à 4 entre 8h et 12h

```python
mesures_caps = matrice_temp[1:4, 8:13]
print(mesures_caps)
```

### Partie 7 – Concaténation 

##### 1) Créer aléatoirement la matrice représentant les températures mesurées par les 3 capteurs durant une journée 

```python
rng=np.random.default_rng(seed=42)
matrice_snd_batiment = rng.integers(20,45,size=(3,24))
print(matrice_snd_batiment)
```

##### 2) Déterminer de deux manières différentes la matrice qui fusionne les températures mesurées par les 8 capteurs durant une journée 

##### Premiere methode :

```python
matrice_fusion_1 = np.vstack((matrice_temp, matrice_snd_batiment))
print(matrice_fusion_1)
```

##### Deuxieme methode :

```python
matrice_fusion_2 = np.concatenate((matrice_temp, matrice_snd_batiment), axis=0)
print(matrice_fusion_2)
```
