### Partie 1 : Listes Python vs tableaux NumPy 

#### 1) Créer une liste Python stockant ces températures; 

```python
import numpy as np
temperatures = [24.5, 25.1, 26.3, 27.0, 25.8 ]
print (temperatures)
```

#### 2) Créer un tableau NumPy équivalent stockant ces températures ;

```python
temperatures = np.array([24.5, 25.1, 26.3, 27.0, 25.8 ])
print(temperatures)
```

#### 3) Effectuer une multiplication par 2 sur ces deux collections; 

---pour la liste---

```python
mult_liste_temp = temperatures * 2
print(mult_liste_temp)
```

---pour le tableau---

```python
mult_tableau_temp = temperatures * 2
print (mult_tableau_temp)
```

#### 4) Quelle différence observe-t-on ? 

Pour la liste : nous constatons qu'il y'a une duplication de la liste ,les valeurs sont repetes deux fois .
Et pour le tableau : nous constatons un nouveau tableau avec des valeurs multipliees chacune par 2.
