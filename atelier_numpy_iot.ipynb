### Partie 1 : Listes Python vs tableaux NumPy 

#### 1) Créer une liste Python stockant ces températures; 

```python
import numpy as np
temperatures = [24.5, 25.1, 26.3, 27.0, 25.8 ]
print (temperatures)
```

#### 2) Créer un tableau NumPy équivalent stockant ces températures ;

```python
temperatures = np.array([24.5, 25.1, 26.3, 27.0, 25.8 ])
print(temperatures)
```

#### 3) Effectuer une multiplication par 2 sur ces deux collections; 

---pour la liste---

```python
mult_liste_temp = temperatures * 2
print(mult_liste_temp)
```

---pour le tableau---

```python
mult_tableau_temp = temperatures * 2
print (mult_tableau_temp)
```

#### 4) Quelle différence observe-t-on ? 

Pour la liste : nous constatons qu'il y'a une duplication de la liste ,les valeurs sont repetes deux fois .
Et pour le tableau : nous constatons un nouveau tableau avec des valeurs multipliees chacune par 2.

### Partie 2 – Création des données 

#### 1) Créer un tableau représentant les mesures de température de 10 capteurs, initialement à zéro. 

```python
tableau_zero = np.zeros(10)
print(tableau_zero)
```

#### 2) Créer un tableau contenant 10 valeurs égales à 1. 

```python
tableau_un = np.ones(10)
print(tableau_un)
```

#### 3) Créer un tableau de 10 valeurs initialisées à 25. 


```python
val = np.full(10,25)
print(val)
```

#### 4) Créer une matrice identité de taille 5 × 5. 

```python
identite = np.eye(5,5)
print(identite)
```

#### 5) Créer les 24 heures (de 0 à 23) de mesure d’une journée 

```python
heures = np.arange(0,24)
print(heures)
```

#### 6) Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C 

```python
tableau_repartie = np.linspace(20,30,10 , dtype=int)
print(tableau_repartie)
```

#### 7) Créer aléatoirement les températures de 100 mesures, en rendant les résultats reproductibles 

```python
rng = np.random.default_rng(seed=42)
temperature = rng.random(100)
print(temperature)
```

**Explication**: ici pour generer aleatoirement les temperatures de 100 mesures en rendant les resultat reproductibles , on initialise une graine "seed" qu'on choisi de fixer ici 42. cette initialisation permet de fixer un point de depart pour que la machine puisse toujours generer les memes 100 temperatures dans le meme ordre sur n'importe quel machine. 

### Partie 3 – Exploration des tableaux 

#### 1) Créer aléatoirement un tableau représentant les 24 températures d'une journée : 

```python
rng = np.random.default_rng(seed=42)
temp_journee = rng.integers(20,45,24)
print(temp_journee)
```

#### 2) Déterminer les propriétés suivantes du tableau 
##### a) nombre de dimensions 

```python
nb_dimensions = temp_journee.ndim
print(nb_dimensions)
```

***explication*** : le resultat indique qu'il y'a une seule dimension c'est a dire une simple liste

##### b) nombre de valeurs ; 

```python
nb_val = temp_journee.size
print(nb_val)
```

***explication*** : ce resultat indique que cette liste contient 24 valeurs

##### c) type des données 

```python
type_donnee = temp_journee.dtype
print(type_donnee)
```

***explication*** : ce resultat indique le type des donnees et nous pouvons constater que les donnees sont de type float.

##### d) forme du tableau

```python
forme_tab = temp_journee.shape
print(forme_tab)
```

***explication*** : ce resultat nous indique qu'il y'a un seule axe contenant 24 elements
